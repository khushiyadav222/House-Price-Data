# Data Documentation

## Dataset Overview
This is a housing dataset containing information about residential properties. The dataset has **1,460 rows** and **81 columns** of property-related features.

## Dataset Structure

### Key Dimensions
- **Total Records**: 1,460 properties
- **Total Features**: 81 columns
- **Data Types**: Mix of numerical (int64, float64) and categorical (object) variables

## Feature Categories

### 1. Property Identification & Location
- `Id`: Unique property identifier
- `MSSubClass`: Building class type
- `MSZoning`: General zoning classification
- `Neighborhood`: Physical location within Ames city limits
- `LotFrontage`, `LotArea`: Lot size measurements
- `LotShape`, `LandContour`: Physical lot characteristics
- `LotConfig`: Lot configuration
- `LandSlope`: Slope of property

### 2. Building Characteristics
- `BldgType`: Type of dwelling
- `HouseStyle`: Style of house
- `YearBuilt`, `YearRemodAdd`: Construction and remodeling years
- `RoofStyle`, `RoofMatl`: Roof characteristics
- `Exterior1st`, `Exterior2nd`: External covering materials
- `OverallQual`, `OverallCond`: Overall quality and condition ratings

### 3. Interior Features
- `1stFlrSF`, `2ndFlrSF`, `GrLivArea`: Floor areas and living space
- `BedroomAbvGr`, `KitchenAbvGr`: Number of bedrooms and kitchens
- `FullBath`, `HalfBath`: Bathroom counts
- `KitchenQual`: Kitchen quality
- `TotRmsAbvGrd`: Total rooms above grade

### 4. Basement Features
- `BsmtQual`, `BsmtCond`: Basement quality and condition
- `BsmtExposure`: Basement exposure level
- `BsmtFinType1`, `BsmtFinType2`: Basement finish types
- `BsmtFinSF1`, `BsmtFinSF2`, `BsmtUnfSF`: Basement area measurements
- `TotalBsmtSF`: Total basement square footage

### 5. Heating & Utilities
- `Heating`: Type of heating system
- `HeatingQC`: Heating quality and condition
- `CentralAir`: Central air conditioning (Y/N)
- `Electrical`: Electrical system type
- `Utilities`: Type of utilities available

### 6. Garage Features
- `GarageType`: Garage location/type
- `GarageYrBlt`: Year garage was built
- `GarageFinish`: Garage interior finish
- `GarageCars`, `GarageArea`: Garage capacity and size
- `GarageQual`, `GarageCond`: Garage quality and condition
- `PavedDrive`: Paved driveway (Y/N)

### 7. Outdoor Features
- `WoodDeckSF`, `OpenPorchSF`, `EnclosedPorch`: Outdoor areas
- `3SsnPorch`, `ScreenPorch`: Seasonal and screened porches
- `PoolArea`, `PoolQC`: Pool features
- `Fence`: Fence quality
- `MiscFeature`, `MiscVal`: Miscellaneous features and value

### 8. Sale Information
- `MoSold`, `YrSold`: Sale month and year
- `SaleType`: Type of sale
- `SaleCondition`: Sale condition
- `SalePrice`: Sale price (target variable)

## Data Quality Notes
- Dataset contains both missing and complete values
- Numerical features include continuous and discrete variables
- Categorical features use text/code values
- Price range varies significantly across properties
- Suitable for real estate price prediction and analysis

## Import Seaborn Library

The `seaborn` library is imported as `sns` to create attractive and informative statistical visualizations. It is built on top of Matplotlib and provides functions for heatmaps, pair plots, violin plots, KDE plots, box plots, and other advanced charts.

```python
import seaborn as sns
```

## Correlation Heatmap

This code creates a correlation heatmap for all numerical columns in the dataset. The `corr(numeric_only=True)` function calculates the correlation matrix, while `sns.heatmap()` visualizes the relationships between variables. The `annot=True` parameter displays the correlation values, and `cmap="coolwarm"` applies a color gradient for better interpretation.

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

## Pair Plot

This code creates a pair plot for the selected columns using Seaborn. A pair plot displays pairwise relationships and the distribution of each variable, helping identify patterns, trends, correlations, and possible outliers in the dataset.

```python
columns = ["Neighborhood", "LandContour", "YearBuilt", "BsmtQual"]

sns.pairplot(df[columns])

plt.show()
```

## Violin Plot

This code creates a violin plot to visualize the distribution of `LotFrontage` across different `LotConfig` categories. The violin plot combines a box plot with a density plot, making it useful for understanding the distribution, spread, and variation of numerical data within each category.

```python
plt.figure(figsize=(10,6))

sns.violinplot(x="LotFrontage", y="LotConfig", data=df)

plt.title("Violin Plot of LotFrontage vs LotConfig")
plt.show()
```

## KDE Plot

This code creates a Kernel Density Estimation (KDE) plot for the `Fireplaces` column after removing missing values using `dropna()`. The KDE plot provides a smooth visualization of the data distribution, making it easier to understand the density and overall pattern of the numerical values.

```python
plt.figure(figsize=(8,5))

sns.kdeplot(df["Fireplaces"].dropna(), fill=True)

plt.title("KDE Plot of Fireplaces")
plt.xlabel("Fireplaces")
plt.show()
```
